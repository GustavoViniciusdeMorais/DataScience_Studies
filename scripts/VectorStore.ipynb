```python
import os
import pypdf
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.embeddings.ollama import OllamaEmbeddings
from langchain.vectorstores.chroma import Chroma
```

```python
# %pip install -qU langchain-ollama
# start linux ollama service
# command: ollama service
```

```python
# Load pdf contents
loader = PyPDFLoader("../pdfs/DDDLaravel.pdf")
docs = loader.load()
```

```python
# Split document in chuncks
text_splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
splits = text_splitter.split_documents(docs)
```

```python
# Load embedding to build document DB
embed = OllamaEmbeddings(model="nomic-embed-text")
```

```python
# Build document DB
abs_path = os.path.abspath("./laravel_ddd")
vector_store = Chroma(
    collection_name="laravel_ddd",
    embedding_function=embed,
    persist_directory=abs_path,  # Where to save data locally, remove if not necessary
)
```

```python
vector_store.add_documents(documents=splits)
# after created, check the contents with sqlite3
# apt update
# apt install sqlite3 -y
# sqlite3 scripts/laravel_ddd
# .tables
# .schema embedding_metadata
# select id, key, string_value from embedding_metadata where id =5;
```
